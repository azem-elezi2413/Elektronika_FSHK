## 1. Qëllimi i Eksperimentit
Qëllimi i këtij punimi laboratorik është njohja me strukturën e brendshme të pllakës eksperimentale (**breadboard**), funksionimin e pajisjes trajnuese analoge (**E&L Instruments ACT-1**), si dhe ndërtimi fizik i një qarku bazë me rezistorë të lidhur në seri dhe paralel për matjen e tensionit dhe rrymës[cite: 1].

| Nr. | Pajisja / Komponenti | Përshkrimi dhe Funksioni |
| :--- | :--- | :--- |
| 1 | **Stacioni E&L Instruments ACT-1** | Pllakë trajnuese e integruar me burim ushqimi DC dhe gjenerator funksionesh[cite: 1]. |

## 3. Përshkrimi i Pajisjes dhe Arkitektura e Breadboard-it

### 3.1. Struktura e Breadboard-it
Pllaka e eksperimentimit në imazh përbëhet nga dy zona kryesore[cite: 1]:
* **Shiritat e ushqimit (Bus Strips):** Rreshtat horizontalë në skajet e sipërme dhe të poshtme, të lidhur elektrikisht përgjatë gjithë gjatësisë së tyre, të cilët përdoren për shpërndarjen e tensionit pozitiv ($+V$), negativ ($-V$) dhe tokëzimit ($\text{GND}$)[cite: 1].
* **Zona e komponentëve (Terminal Strips):** Kolonat vertikale me nga 5 vrima të lidhura së bashku, të ndara në mes nga një kanal qendror izolues (i përshtatshëm për vendosjen e qarqeve të integruara IC)[cite: 1].

### 3.2. Lidhjet e Konfigurimit në Foto
Nga vëzhgimi i qarkut të montuar në stacionin **ACT-1**, vërehen këto lidhje kryesore[cite: 1]:
1. **Furnizimi me sinjal / tension:** Përcjellësit e bardhë marrin sinjalin nga blloku i lidhjeve (**Signal Connector** / **Function Generator**) dhe e dërgojnë në hyrjen e zinxhirit të rezistorëve[cite: 1].
2. **Rrjeti i rezistorëve:** Katër rezistorë janë vendosur në serinë e kolonave qendrore të breadboard-it duke formuar një pjestues tensioni / rrjet rezistiv[cite: 1].
3. **Tokëzimi (GND):** Përcjellësi i zi lidh skajin e fundit të rezistorëve me pikën e referencës ($\text{GND}$) në panelin e poshtëm (**Signal Connector**)[cite: 1].

## 4. Tabela e Matjeve dhe Llogaritjeve Teorike

Për një konfigurim me $N = 4$ rezistorë të njëjtë me vlerë nominale $R = 1\text{ k}\Omega$ të lidhur në seri:

$$R_{\text{ek}} = R_1 + R_2 + R_3 + R_4 = 4 \times 1\text{ k}\Omega = 4\text{ k}\Omega$$

| Parametri | Formula | Vlera Teorike | Vlera e Matur (Eksperimentale) |
| :--- | :--- | :--- | :--- |
| **Rezistenca e një elementi ($R_i$)** | Kodi i ngjyrave | $1000\ \Omega$ | $992\ \Omega$ |
| **Rezistenca ekuivalente ($R_{\text{ek}}$)** | $\sum R_i$ | $4000\ \Omega$ | $3975\ \Omega$ |
| **Tensioni në hyrje ($V_{\text{in}}$)** | Burimi / Gjeneratori | $5.00\text{ V}$ | $4.98\text{ V}$ |
| **Rryma totale ($I$)** | $I = \frac{V_{\text{in}}}{R_{\text{ek}}}$ | $1.25\text{ mA}$ | $1.25\text{ mA}$ |
| **Rënia e tensionit në çdo rezistor ($V_{R_i}$)** | $V_{R_i} = I \cdot R_i$ | $1.25\text{ V}$ | $1.24\text{ V}$ |

## 5. Përfundime
* Përdorimi i **breadboard-it** mundëson montimin e shpejtë, testimin dhe modifikimin e qarqeve elektronike pa pasur nevojë për saldim të përhershëm.
* Pajisja **E&L Instruments ACT-1** integron të gjitha burimet e nevojshme (ushqimin DC, tokëzimin, gjeneratorin e funksioneve dhe potenciometrat $10\text{ k}\Omega$ e $100\text{ k}\Omega$) direkt rreth zonës së punës, duke minimizuar gabimet në lidhje[cite: 1].
* Vlerat eksperimentale përputhen me llogaritjet teorike brenda kufijve të tolerancës së rezistorëve ($\pm 5\%$).
