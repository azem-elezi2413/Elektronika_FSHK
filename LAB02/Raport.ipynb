## 1. Qellimi i Eksperimentit
Qellimi i ketij punimi laboratorik eshte studimi dhe verifikimi eksperimental i skemave te pjestuesit te tensionit (voltage divider) dhe sjelljes se qarqeve elektrike kur ne to lidhet nje rezistence ngarkese ($R_L$), duke krahasuar vlerat e matura me formulat teorike .

---

## 2. Te dhenat dhe Rezultatet e Eksperimenteve

### Eksperimenti 1: Pjestuesi Baze i Tensionit
Ne kete eksperiment u studiua qarku pjestues i tensionit i perbere nga dy rezistenca ne seri dhe nje burim tensioni ne hyrje .

* **Vlerat e matura dhe nominale:**
  - $V_{in} = 15,35\text{ V}$ 
  - $V_{out} = 7,5\text{ V}$ 
  - $R_1 = 7,55\text{ k}\Omega$ 
  - $R_2 = 7,78\text{ k}\Omega$ 
* **Formula e zbatimit:**
  $$\frac{V_{out}}{V_{in}} = \frac{R_2}{R_1 + R_2}$$
* Vlerat tregojne se pjestimi i tensionit ne nyjen e mesit ndermjet dy rezistencave perkon saktesisht me raportin e rezistencave perkatese .

---

### Eksperimenti 2: Qarku i Hapur (Pa Ngarkese)
U analizua qarku me rezistenca me vlera te larta (ne rangun e megaohmave) ne gjendje te hapur .

* **Parametrat e matur:**
  - $R_1 = 3,27\text{ M}\Omega$ 
  - $R_2 = 3,30\text{ M}\Omega$ 
  - $V_{in} = 6,82\text{ V}$ (ose $V = 15,77\text{ V}$ ne total te burimit) 
  - $V_{out} = 6,71\text{ V}$ 

---

### Eksperimenti 3: Qarku me Rezistence Ngarkese ($R_L$)
U studiua ndikimi i lidhjes se nje rezistence ngarkese ($R_L$) ne dalje te pjestuesit te tensionit . Rezistenca e ngarkeses lidhet ne paralel me rezistencen $R_2$.

* **Formula e ekuivalencës për rezistencat në paralel:**
  $$R_2 \parallel R_L = \frac{R_2 \cdot R_L}{R_2 + R_L}$$
* **Formula e tensionit ne dalje me ngarkese:**
  $$V_{out} = \frac{R_2 \parallel R_L}{R_1 + R_2 \parallel R_L} \cdot V_{in}$$

* **Rasti 1:**
  - $R_1 = 7,55\text{ k}\Omega$ 
  - $R_2 = 7,78\text{ k}\Omega$ 
  - $R_L = 3,30\text{ M}\Omega$ 
  - Tensioni i matur $V = 7,76\text{ V}$ 

* **Rasti 2:**
  - $R_1 = 7,55\text{ k}\Omega$ 
  - $R_2 = 7,78\text{ k}\Omega$ 
  - $R_L = 2,34\text{ M}\Omega$ 
  - Tensioni i matur $V = 7,71\text{ V}$ 

---

## 3. Perfundime
* Rezultatet eksperimentale konfirmojne formula teorike te pjestuesit te tensionit.
* Prania e rezistences se ngarkeses ($R_L$) ndryshon rezistencen ekuivalente te deges se poshtme ($R_2 \parallel R_L$), gje qe con ne nje ulje te vleres se tensionit ne dalje ($V_{out}$) krahasuar me gjendjen pa ngarkese.
* Matjet e kryera ne laborator perputhen me llogaritjet teorike, duke marre parasysh tolerancat e komponentëve dhe rezistencën e brendshme të instrumenteve matëse .
